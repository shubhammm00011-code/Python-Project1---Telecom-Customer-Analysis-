```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
```

```python
df = pd.read_csv('Customer Churn.csv')
df.info()
```

#here we are replacing blanks with "0" as tenure is zero and total charges are not recorded and converted the datatype from obj to float.

```python
df["TotalCharges"] = df["TotalCharges"].replace(" ", "0")
df["TotalCharges"] = df["TotalCharges"].astype("float")
```

```python
df.info()
```

#checking null values

```python
df.isnull().sum()
```

```python
df.describe()
```

```python
df["customerID"].duplicated().sum()
```

```python
def conv(value):
    if value == 1:
        return "yes"
    else:
        return "no"

df["SeniorCitizen"]= df["SeniorCitizen"].apply(conv)        
    
    
```

```python
df.info()
```

```python
df.head()
```

```python
ax = sns.countplot(x = 'Churn', data = df)

ax.bar_label(ax.containers[0])
plt.title("Count of Churn")
plt.show()
```

```python
plt.figure(figsize = (3,3))
gb = df.groupby("Churn").agg({'Churn':"count"})
plt.pie(gb['Churn'], labels = gb.index , autopct = "%1.2f%%")
plt.title("Percentage of Churned Customer", fontsize = 20)
plt.show()
```

#from the given pie chart above , about 26.54% customers are churned out and lets know the reason behind it.

```python
plt.figure(figsize=(5,4))
ax = sns.countplot( x = 'gender' , data = df , hue = "Churn")
ax.bar_label(ax.containers[0])
plt.title("Impact of Gender on Customer Churn" ,fontsize = 20)
plt.show()
```

```python
plt.figure(figsize = (4,4))
ax = sns.countplot(x = "SeniorCitizen", data = df , hue="Churn")
ax.bar_label(ax.containers[0])
plt.title("Count of Customers by Senior Citizen")
plt.show()
```

```python
total_counts = df.groupby('SeniorCitizen')['Churn'].value_counts(normalize=True).unstack() * 100

# Plot
fig, ax = plt.subplots(figsize=(4, 4))  # Adjust figsize for better visualization

# Plot the bars
total_counts.plot(kind='bar', stacked=True, ax=ax, color=['#1f77b4', '#ff7f0e'])  # Customize colors if desired

# Add percentage labels on the bars
for p in ax.patches:
    width, height = p.get_width(), p.get_height()
    x, y = p.get_xy()
    ax.text(x + width / 2, y + height / 2, f'{height:.1f}%', ha='center', va='center')

plt.title('Churn by Senior Citizen (Stacked Bar Chart)')
plt.xlabel('SeniorCitizen')
plt.ylabel('Percentage (%)')
plt.xticks(rotation=0)
plt.legend(title='Churn', bbox_to_anchor = (0.9,0.9))  # Customize legend location

plt.show()
```

```python
# 1. Define the list of columns you want to plot
service_columns = [
    'PhoneService', 'MultipleLines', 'InternetService', 
    'OnlineSecurity', 'OnlineBackup', 'DeviceProtection', 
    'TechSupport', 'StreamingTV', 'StreamingMovies'
]

# 2. Create a 3x3 grid of subplots
# figsize=(15, 15) ensures the charts are large enough to read easily
fig, axes = plt.subplots(nrows=3, ncols=3, figsize=(15, 15))

# Flatten the 3x3 axes array into a 1D list so we can loop through it easily
axes = axes.flatten()

# 3. Loop through the columns and the axes
for i, col in enumerate(service_columns):
    # Create the countplot on the specific axis (ax=axes[i])
    sns.countplot(data=df, x=col, hue='Churn', ax=axes[i])
    
    # Add a title to each subplot to keep it organized
    axes[i].set_title(f"Churn vs {col}", fontsize=12, fontweight='bold')
    
    # Optional: Remove the X-axis label since the title already explains it
    axes[i].set_xlabel('')

# 4. Adjust layout so the plots and labels don't overlap
plt.tight_layout()

# Display the final dashboard
plt.show()
```

#The majority of customers who do not churn tend to have services like PhoneService, InternetService (particularly DSL), and OnlineSecurity enabled. For services like OnlineBackup, TechSupport, and StreamingTV, churn rates are noticeably higher when these services are not used or are unavailable.
